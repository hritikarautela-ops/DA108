```python
import time
import random
import numpy as np
print("Numpy Version:", np.__version__)
```

```python
a_list = [1, 2, 3, 4]
b_list = [10, 20, 30, 40]
a = np.array(a_list)
b = np.array(b_list)
y_np = a * b
print("Result:", y_np.tolist())
```

```python
# matrix vector multiplication
a_list = [[1, 2, 3],
          [4, 5, 6]]
x_list = [10, 20, 30]

 # Without numpy
def matvec(A,x):
  m = len(A)
  n = len(x)
  y = [0.0] * m
  for i in range(m):
    row = A[i]
    s = 0.0
    for j in range(n):
      s += row[j] * x[j]
      y[i] = s
  return y


y_loop = matvec(a_list, x_list)
print("Result:", y_loop)

# With numpy

A = np.array(a_list, dtype = float)
X = np.array(x_list, dtype = float)
#y_np = np.dot(A, X)
y_np = A @ X
print("Result:", y_np.tolist())
```

```python
elem = A * X
print("A Shape:", A.shape)
print("X Shape:", X.shape)
print("A * X Shape:", elem.shape)
print("Elementwise multiplication:\n", elem)
```

```python
A = np.random.rand(5, 4)
B = np.array([ 1.0, 10.0, 100.0, 1000.0])
out = A * B
print("A Shape:", A.shape)
print("B Shape:", B.shape)
print("Out Shape:", out.shape)
print("First row before:", A[0])
print("First row after:", out[0])
```

```python
def time_block(fn, repeats = 5):
  times = []
  for _ in range(repeats):
    t0 = time.perf_counter()
    fn()
    t1 = time.perf_counter()
    times.append(t1 - t0)
  return min(times), sum(times)/len(times)

m, n = 400, 400
A_list = [[random.random() for _ in range(n)] for _ in range(m)]
x_list = [random.random()for _ in range(n)]

A_np = np.array(A_list, dtype = np.float64)
x_np = np.array(x_list, dtype = np.float64)

_ = A_np @ x_np

best_loop, avg_loop = time_block(lambda: matvec(A_list, x_list))
best_np, avg_np = time_block(lambda: A_np @ x_np)

print(f"Best matvec: {best_loop:.6f}s, avg:{avg_loop:.6f}")
print(f"Best numpy: {best_np:.6f}s, avg:{avg_np:.6f}")
```
